# Phân tích dữ liệu

## Câu hỏi nghiên cứu

Lặp lại các đề mục bên dưới cho từng câu hỏi. Ghi người phụ trách, động cơ và tiêu chí độ sâu (D1–D6) áp dụng.

### Tiền xử lý

Mô tả từng bước, kiểm tra số dòng trước và sau, và giải thích thay đổi.

### Phương pháp

Trình bày phương pháp phân tích và lý do lựa chọn.

### Kết quả

Thêm ít nhất hai biểu đồ có nhãn đầy đủ. Trả lời câu hỏi bằng số liệu và thảo luận ý nghĩa thực tế.

## Tổng kết và phản ánh

Tổng hợp kết quả, giới hạn và hướng phát triển. Mỗi thành viên tự viết phần phản ánh cá nhân.
